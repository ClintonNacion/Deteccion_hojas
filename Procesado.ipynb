```python
import cv2
import os
import shutil
import numpy as np

# Rutas
input_folder = "/content/drive_mount/MyDrive/Capstone/hoja_sana"
output_folder = "/content/drive_mount/MyDrive/Capstone/hoja_sana_procesadas"

# Crear carpeta de salida limpia
if os.path.exists(output_folder):
    shutil.rmtree(output_folder)
os.makedirs(output_folder)

# Parámetros
blur_threshold = 40.0
target_size = (224, 224)

# Función para verificar borrosidad
def is_blurry(image, threshold=blur_threshold):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    lap_var = cv2.Laplacian(gray, cv2.CV_64F).var()
    return lap_var < threshold

# Función para redimensionar con padding (mantiene aspecto)
def resize_with_padding(image, size=(224, 224)):
    h, w = image.shape[:2]
    scale = min(size[0]/h, size[1]/w)
    new_w, new_h = int(w * scale), int(h * scale)
    resized = cv2.resize(image, (new_w, new_h))

    delta_w = size[1] - new_w
    delta_h = size[0] - new_h
    top, bottom = delta_h // 2, delta_h - (delta_h // 2)
    left, right = delta_w // 2, delta_w - (delta_w // 2)

    # Padding con color negro
    color = [0, 0, 0]
    new_image = cv2.copyMakeBorder(resized, top, bottom, left, right, cv2.BORDER_CONSTANT, value=color)
    return new_image

# Procesamiento
count = 0
for filename in os.listdir(input_folder):
    file_path = os.path.join(input_folder, filename)
    image = cv2.imread(file_path)

    if image is None:
        continue

    if not is_blurry(image):
        padded = resize_with_padding(image, target_size)
        save_path = os.path.join(output_folder, filename)
        cv2.imwrite(save_path, padded)
        count += 1

print(f"✅ Procesamiento completo Hojas Sanas. Imágenes guardadas: {count}")
```

```python
import os
import shutil
import random

random.seed(42)

# Configuración de rutas
base_propia = "/content/drive_mount/MyDrive/Capstone/dataset_propia"
base_kaggle = "/content/drive_mount/MyDrive/Capstone/dataset_kaggle"
dest_base = "/content/drive_mount/MyDrive/Capstone/dataset"

# Clases y su proporción
clases = ["healthy", "leafminer", "phoma", "rust"]

for clase in clases:
    print(f"\n🔄 Procesando clase: {clase}")

    # Obtener archivos
    propia_imgs = [os.path.join(base_propia, clase, f) for f in os.listdir(os.path.join(base_propia, clase)) if f.endswith('.jpg')]
    kaggle_imgs = [os.path.join(base_kaggle, clase, f) for f in os.listdir(os.path.join(base_kaggle, clase)) if f.endswith('.jpg')]

    total_imgs = propia_imgs + kaggle_imgs
    random.shuffle(total_imgs)

    # Calcular cortes
    total_count = len(total_imgs)
    train_count = int(total_count * 0.8)

    train_imgs = total_imgs[:train_count]
    val_imgs = total_imgs[train_count:]

    # Función para mover
    def mover(imagenes, destino):
        for img in imagenes:
            filename = os.path.basename(img)
            dest_path = os.path.join(dest_base, destino, clase, filename)
            shutil.copy(img, dest_path)

    mover(train_imgs, "train")
    mover(val_imgs, "val")

    print(f"📁 Train: {len(train_imgs)} imágenes")
    print(f"📁 Val: {len(val_imgs)} imágenes")

print("\n✅ División completada con éxito.")
```

##Oversampling y data augmentación

```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
import numpy as np

base_train = "/content/drive_mount/MyDrive/Capstone/dataset/train"
clases = ["healthy", "leafminer", "phoma", "rust"]

# Conteo actualizado de imágenes por clase (Train)
conteo = {
    "healthy": 458,
    "leafminer": 153,
    "phoma": 153,
    "rust": 223
}

max_count = max(conteo.values())

datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    fill_mode='nearest'
)

for clase in clases:
    if conteo[clase] == max_count:
        continue  # No necesitamos aumentar esta clase

    carpeta = os.path.join(base_train, clase)
    imgs = [f for f in os.listdir(carpeta) if f.endswith('.jpg')]
    n_imgs = len(imgs)
    faltan = max_count - n_imgs
    print(f"\nClase {clase} tiene {n_imgs} imágenes. Generando {faltan} aumentadas...")

    # Cuántas nuevas imágenes generar por cada imagen original
    aug_por_img = faltan // n_imgs
    extra = faltan % n_imgs

    for i, img_file in enumerate(imgs):
        img_path = os.path.join(carpeta, img_file)
        img = load_img(img_path)
        x = img_to_array(img)
        x = x.reshape((1,) + x.shape)

        total_aug = aug_por_img + (1 if i < extra else 0)
        count = 0

        for batch in datagen.flow(
            x,
            batch_size=1,
            save_to_dir=carpeta,
            save_prefix=f"aug_{i}",
            save_format='jpg'
        ):
            count += 1
            if count >= total_aug:
                break

print("\n✅ Aumento de datos balanceado y diverso completado.")
```

##MobileNetV2

```python
# === 2. Importar librerías ===
import os
import pickle
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau

# === 3. Configuración ===
train_dir = "/content/drive/MyDrive/Capstone/dataset/train"
val_dir = "/content/drive/MyDrive/Capstone/dataset/val"

save_dir = "/content/drive/MyDrive/Capstone/modelos"
os.makedirs(save_dir, exist_ok=True)

modelo_path = os.path.join(save_dir, "mejor_modelo_mobilenetv2.keras")
historial_path = os.path.join(save_dir, "history_mobilenetv2.pkl")

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 20
FINE_TUNE_EPOCHS = 20
NUM_CLASSES = 4

# === 4. Data augmentation ===
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)

val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# === 5. Modelo base ===
base_model = MobileNetV2(input_shape=(*IMG_SIZE, 3),
                         include_top=False,
                         weights='imagenet')
base_model.trainable = False

# === 6. Modelo completo ===
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.6),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# === 7. Callbacks (con rutas en Drive) ===
checkpoint_cb = ModelCheckpoint(
    filepath=modelo_path,
    save_best_only=True,
    monitor="val_accuracy",
    mode="max",
    verbose=1
)

earlystop_cb = EarlyStopping(
    monitor="val_loss",
    patience=6,
    mode="min",
    restore_best_weights=True,
    verbose=1
)

reduce_lr_cb = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    verbose=1,
    mode="min",
    min_lr=1e-7
)

# === 8. Entrenamiento inicial ===
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb]
)

# === 9. Fine-tuning ===
base_model.trainable = True
fine_tune_at = 40

for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

total_epochs = EPOCHS + FINE_TUNE_EPOCHS

history_fine = model.fit(
    train_generator,
    epochs=total_epochs,
    initial_epoch=history.epoch[-1],
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb]
)

# === 10. Guardar historial en Drive ===
with open(historial_path, "wb") as f:
    pickle.dump({'initial': history.history, 'fine_tune': history_fine.history}, f)

print("✅ Entrenamiento completado y todo guardado en tu Google Drive.")
```

```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input as preprocess_resnet
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import pickle

# === 1. Rutas ===
train_dir = "/content/drive/MyDrive/Capstone/dataset/train"
val_dir = "/content/drive/MyDrive/Capstone/dataset/val"
save_dir = "/content/drive/MyDrive/Capstone/modelos"
os.makedirs(save_dir, exist_ok=True)

modelo_path = os.path.join(save_dir, "mejor_modelo_resnet50.keras")
historial_path = os.path.join(save_dir, "history_resnet.pkl")

# === 2. Parámetros ===
IMG_SIZE = (240, 240)
BATCH_SIZE = 32
EPOCHS = 20
FINE_TUNE_EPOCHS = 20
NUM_CLASSES = 4

# === 3. Data augmentation ===
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_resnet,
    rotation_range=30,
    zoom_range=0.2,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.15,
    brightness_range=[0.8, 1.2],
    channel_shift_range=30.0,
    horizontal_flip=True,
    fill_mode='nearest'
)

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_resnet)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)

val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# === 4. Class weights (balanceo) ===
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_generator.classes),
    y=train_generator.classes
)
class_weights = dict(enumerate(class_weights))

# === 5. Cargar base ResNet50 ===
base_model = ResNet50(
    input_shape=(*IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False

# === 6. Modelo completo ===
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.BatchNormalization(),
    layers.Dense(256, activation='relu', kernel_regularizer=regularizers.l2(0.0005)),
    layers.BatchNormalization(),
    layers.Dropout(0.4),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# === 7. Callbacks ===
checkpoint_cb = ModelCheckpoint(
    filepath=modelo_path,
    save_best_only=True,
    monitor="val_accuracy",
    mode="max",
    verbose=1
)

earlystop_cb = EarlyStopping(
    monitor="val_accuracy",
    patience=6,
    mode="max",
    restore_best_weights=True,
    verbose=1
)

reduce_lr_cb = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    verbose=1,
    mode="min",
    cooldown=1,
    min_delta=0.001,
    min_lr=1e-7
)

# === 8. Entrenamiento inicial ===
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb],
    class_weight=class_weights
)

# === 9. Fine-tuning: descongela las últimas 50 capas ===
base_model.trainable = True
fine_tune_at = len(base_model.layers) - 50
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

# Cambiar optimizador a SGD
model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=1e-4, momentum=0.9),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

total_epochs = EPOCHS + FINE_TUNE_EPOCHS

history_fine = model.fit(
    train_generator,
    epochs=total_epochs,
    initial_epoch=history.epoch[-1] + 1,
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb],
    class_weight=class_weights
)

# === 10. Guardar historial ===
with open(historial_path, "wb") as f:
    pickle.dump({'initial': history.history, 'fine_tune': history_fine.history}, f)

print("✅ Modelo optimizado ResNet50 entrenado y guardado con éxito.")

```

```python
import os
import pickle
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.applications.inception_v3 import preprocess_input as preprocess_incep
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.models import load_model

# === Configuración de rutas ===
train_dir = "/content/drive/MyDrive/Capstone/dataset/train"
val_dir = "/content/drive/MyDrive/Capstone/dataset/val"
save_dir = "/content/drive/MyDrive/Capstone/modelos"
os.makedirs(save_dir, exist_ok=True)

modelo_path = os.path.join(save_dir, "mejor_modelo_inceptionv3.keras")
historial_path = os.path.join(save_dir, "history_inceptionv3.pkl")

# === Parámetros ===
IMG_SIZE = (299, 299)
BATCH_SIZE = 32
EPOCHS = 20
FINE_TUNE_EPOCHS = 20
NUM_CLASSES = 4

# === Data augmentation ===
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_incep,
    rotation_range=25,
    zoom_range=0.25,
    width_shift_range=0.15,
    height_shift_range=0.15,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_incep)

train_generator = train_datagen.flow_from_directory(
    train_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)

val_generator = val_datagen.flow_from_directory(
    val_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

# === Modelo base (congelado al inicio) ===
base_model = InceptionV3(
    input_shape=(*IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False

# === Clasificador superior ===
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.BatchNormalization(),
    layers.Dense(256, activation='relu', kernel_regularizer=regularizers.l2(1e-4)),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

model.compile(
    optimizer=tf.keras.optimizers.RMSprop(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# === Callbacks ===
checkpoint_cb = ModelCheckpoint(
    filepath=modelo_path,
    save_best_only=True,
    monitor="val_accuracy",
    mode="max",
    verbose=1
)

earlystop_cb = EarlyStopping(
    monitor="val_accuracy",
    patience=6,
    restore_best_weights=True,
    verbose=1,
    mode="max"
)

reduce_lr_cb = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    min_lr=1e-7,
    verbose=1
)

# === Entrenamiento inicial ===
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb]
)

# === 🔁 Cargar el mejor modelo antes de fine-tuning ===
model = load_model(modelo_path)
print("✅ Modelo recargado desde el mejor checkpoint.")

# === Descongelar parte del modelo base ===
base_model = model.layers[0]
base_model.trainable = True

fine_tune_at = 150  # o 100, 200 según tu prueba
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

# === Compilar con menor learning rate ===
model.compile(
    optimizer=tf.keras.optimizers.RMSprop(learning_rate=1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# === Fine-tuning desde 0 (nuevo historial) ===
history_fine = model.fit(
    train_generator,
    epochs=FINE_TUNE_EPOCHS,
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb]
)

# === Guardar historial combinado ===
with open(historial_path, "wb") as f:
    pickle.dump({
        'initial': history.history,
        'fine_tune': history_fine.history
    }, f)

print("✅ Fine-tuning completado desde el mejor modelo y todo guardado en Drive.")
```

```python
# === 2. Importar librerías ===
import os
import pickle
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV3Small
from tensorflow.keras.applications.mobilenet_v3 import preprocess_input
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from tensorflow.keras import regularizers

# === 3. Configuración ===
train_dir = "/content/drive/MyDrive/Capstone/dataset/train"
val_dir = "/content/drive/MyDrive/Capstone/dataset/val"

save_dir = "/content/drive/MyDrive/Capstone/modelos"
os.makedirs(save_dir, exist_ok=True)

modelo_path = os.path.join(save_dir, "mejor_modelo_mobilenetv3.keras")
historial_path = os.path.join(save_dir, "history_mobilenetv3.pkl")

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 20
FINE_TUNE_EPOCHS = 20
NUM_CLASSES = 4

# === 4. Data augmentation ===
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True
)

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)

val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# === 5. Modelo base ===
base_model = MobileNetV3Small(
    input_shape=(*IMG_SIZE, 3),
    include_top=False,
    weights='imagenet',
    dropout_rate=0.2
)
base_model.trainable = False

# === 6. Modelo completo ===
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(0.015)),  # L2 más fuerte
    layers.Dropout(0.5),  # Dropout más alto
    layers.Dense(NUM_CLASSES, activation='softmax')
])


model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# === 7. Callbacks ===
checkpoint_cb = ModelCheckpoint(
    filepath=modelo_path,
    save_best_only=True,
    monitor="val_accuracy",
    mode="max",
    verbose=1
)

earlystop_cb = EarlyStopping(
    monitor="val_loss",
    patience=6,
    mode="min",
    restore_best_weights=True,
    verbose=1
)

reduce_lr_cb = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3,
    verbose=1,
    mode="min",
    min_lr=1e-7
)

# === 8. Entrenamiento inicial ===
history = model.fit(
    train_generator,
    epochs=EPOCHS,
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb]
)

# === 9. Fine-tuning ===
# Descongelar solo las últimas 50 capas
for layer in base_model.layers[:-50]:
    layer.trainable = False
for layer in base_model.layers[-50:]:
    if not isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = True


model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

total_epochs = EPOCHS + FINE_TUNE_EPOCHS

history_fine = model.fit(
    train_generator,
    epochs=total_epochs,
    initial_epoch=history.epoch[-1],
    validation_data=val_generator,
    callbacks=[checkpoint_cb, earlystop_cb, reduce_lr_cb]
)

# === 10. Guardar historial ===
with open(historial_path, "wb") as f:
    pickle.dump({'initial': history.history, 'fine_tune': history_fine.history}, f)

print("✅ Entrenamiento completado y modelo MobileNetV3Large guardado.")
```

##ConvNeXt-Tiny

```python
import os, torch, timm, pickle, copy
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from torch.optim.lr_scheduler import ReduceLROnPlateau

# ───── Configuración ─────
device      = torch.device("cuda" if torch.cuda.is_available() else "cpu")
train_dir   = "/content/drive/MyDrive/Capstone/dataset/train"
val_dir     = "/content/drive/MyDrive/Capstone/dataset/val"
batch_size  = 32
img_size    = 224
num_classes = 4
epochs      = 30                           # ↑ damos margen, EarlyStopping lo recorta
save_dir    = "/content/drive/MyDrive/Capstone/modelos"
os.makedirs(save_dir, exist_ok=True)

# ───── Transformaciones ─────
train_tfms = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(25),
    transforms.ColorJitter(0.2, 0.2, 0.2, 0.1),
    transforms.RandomResizedCrop(img_size, scale=(0.8, 1.0)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

val_tfms = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# ───── Datasets ─────
train_ds = datasets.ImageFolder(train_dir, transform=train_tfms)
val_ds   = datasets.ImageFolder(val_dir,   transform=val_tfms)
train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=batch_size, shuffle=False)

# ───── Modelo con DropPath ─────
model = timm.create_model(
    'convnext_tiny',
    pretrained=True,
    num_classes=num_classes,
    drop_path_rate=0.3          # ← regularización tipo Dropout
).to(device)

criterion  = nn.CrossEntropyLoss()
optimizer  = torch.optim.Adam(model.parameters(), lr=1e-4, weight_decay=1e-3)
scheduler  = ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=3, verbose=True)

# ───── EarlyStopping ─────
class EarlyStopping:
    def __init__(self, patience=5):
        self.patience = patience
        self.counter = 0
        self.best_acc = None
        self.best_model = None
        self.stop = False

    def __call__(self, val_acc, model):
        if self.best_acc is None or val_acc > self.best_acc:
            self.best_acc = val_acc
            self.best_model = copy.deepcopy(model.state_dict())
            self.counter = 0
        else:
            self.counter += 1
            print(f"⏳ EarlyStopping {self.counter}/{self.patience}")
            if self.counter >= self.patience:
                self.stop = True

early_stopper = EarlyStopping(patience=5)

# ───── Historial ─────
train_losses, val_losses = [], []
train_accs,   val_accs   = [], []

def evaluate():
    model.eval()
    loss_tot, correct = 0.0, 0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            out  = model(x)
            loss_tot += criterion(out, y).item() * x.size(0)
            correct += (out.argmax(1) == y).sum().item()
    return loss_tot / len(val_ds), correct / len(val_ds)

# ───── Entrenamiento ─────
best_val_acc = 0.0
for epoch in range(1, epochs + 1):
    model.train()
    run_loss, correct = 0.0, 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        out  = model(x)
        loss = criterion(out, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        run_loss += loss.item() * x.size(0)
        correct  += (out.argmax(1) == y).sum().item()

    tr_loss = run_loss / len(train_ds)
    tr_acc  = correct  / len(train_ds)
    val_loss, val_acc = evaluate()

    train_losses.append(tr_loss);  val_losses.append(val_loss)
    train_accs.append(tr_acc);     val_accs.append(val_acc)

    print(f"Epoch {epoch:02}/{epochs} | "
          f"Train Loss {tr_loss:.4f} | Train Acc {tr_acc:.4f} | "
          f"Val Loss {val_loss:.4f} | Val Acc {val_acc:.4f}")

    scheduler.step(val_acc)
    early_stopper(val_acc, model)

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), os.path.join(save_dir, "mejor_modelo_convnext_tiny.pt"))
        print(f"✅ Mejor modelo guardado (Val Acc {val_acc:.4f})")

    if early_stopper.stop:
        print("⛔ Entrenamiento detenido por EarlyStopping.")
        model.load_state_dict(early_stopper.best_model)
        break

# ───── Guardar historial ─────
with open(os.path.join(save_dir, "convnext_tiny_history.pkl"), "wb") as f:
    pickle.dump({
        "train_loss": train_losses,
        "val_loss":   val_losses,
        "train_accuracy": train_accs,
        "val_accuracy":   val_accs
    }, f)
print("📦 Historial guardado.")


```

```python
import os, torch, timm, pickle, copy
import torch.nn as nn
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from torch.optim.lr_scheduler import ReduceLROnPlateau

# ─────────────────────────── Configuración ────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
train_dir = "/content/drive/MyDrive/Capstone/dataset/train"
val_dir  = "/content/drive/MyDrive/Capstone/dataset/val"
save_dir = "/content/drive/MyDrive/Capstone/modelos"
os.makedirs(save_dir, exist_ok=True)

batch_size = 32
img_size   = 224
num_classes = 4
epochs = 40
best_model_path = os.path.join(save_dir, "mejor_modelo_efficientvit_b1.pt")
pkl_path       = os.path.join(save_dir, "efficientvit_b1_history.pkl")

# ─────────────────────────── Transformaciones ─────────────────────────
train_tfms = transforms.Compose([
    transforms.RandomResizedCrop(img_size, scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(30),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
    transforms.RandomAffine(degrees=0, translate=(0.15, 0.15), shear=10),
    transforms.ToTensor(),
    transforms.RandomErasing(p=0.25),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

val_tfms = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

train_ds = datasets.ImageFolder(train_dir, transform=train_tfms)
val_ds   = datasets.ImageFolder(val_dir,   transform=val_tfms)
train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
val_loader   = DataLoader(val_ds,   batch_size=batch_size, shuffle=False)

# ───────────────────────────── Modelo ─────────────────────────────────
model = timm.create_model(
    "efficientvit_b1",
    pretrained=True,
    num_classes=num_classes,
    drop_rate=0.4              # ← Dropout para reducir overfitting
).to(device)

criterion  = nn.CrossEntropyLoss(label_smoothing=0.1)  # ← Label Smoothing
optimizer = torch.optim.Adam(model.parameters(), lr=5e-5, weight_decay=1e-4)  # 🔸 lr más bajo

scheduler  = ReduceLROnPlateau(
    optimizer,
    mode='max',
    factor=0.5,
    patience=3,
    verbose=True,
    min_lr=1e-7
)


# ────────────────────────── EarlyStopping ─────────────────────────────
class EarlyStopping:
    def __init__(self, patience=5, verbose=True):
        self.patience = patience
        self.counter = 0
        self.best_score = None
        self.early_stop = False
        self.verbose = verbose
        self.best_model_wts = None

    def __call__(self, val_acc, model):
        if self.best_score is None or val_acc > self.best_score:
            self.best_score = val_acc
            self.counter = 0
            self.best_model_wts = copy.deepcopy(model.state_dict())
        else:
            self.counter += 1
            if self.verbose:
                print(f"⏳ EarlyStopping contador: {self.counter}/{self.patience}")
            if self.counter >= self.patience:
                self.early_stop = True


# ───────────────────── Función evaluación validación ──────────────────
def evaluate_val():
    """Devuelve (accuracy, loss) sobre validation set."""
    model.eval()
    correct, total_loss = 0, 0.0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            outputs = model(x)
            loss = criterion(outputs, y)
            total_loss += loss.item() * x.size(0)
            preds = outputs.argmax(dim=1)
            correct += (preds == y).sum().item()
    return correct / len(val_ds), total_loss / len(val_ds)

# ────────────────────────── Entrenamiento ─────────────────────────────
train_losses, val_losses = [], []
train_accuracies, val_accuracies = [], []

best_acc = 0.0
early_stopper = EarlyStopping(patience=5)

for epoch in range(epochs):
    # ---- entrenamiento ----
    model.train()
    running_loss, correct_train = 0.0, 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        preds = model(x)
        loss = criterion(preds, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * x.size(0)
        correct_train += (preds.argmax(dim=1) == y).sum().item()

    avg_train_loss = running_loss / len(train_ds)
    train_acc = correct_train / len(train_ds)

    # ---- validación ----
    val_acc, val_loss = evaluate_val()

    # ---- registrar métricas ----
    train_losses.append(avg_train_loss)
    val_losses.append(val_loss)
    train_accuracies.append(train_acc)
    val_accuracies.append(val_acc)

    # ---- log ----
    print(f"📈 Epoch {epoch+1}/{epochs} | "
          f"Train Loss: {avg_train_loss:.4f} | Train Acc: {train_acc:.4f} | "
          f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.4f}")

    # ---- scheduler & early stopping ----
    scheduler.step(val_acc)
    early_stopper(val_acc, model)

    if val_acc > best_acc:
        best_acc = val_acc
        torch.save(model.state_dict(), best_model_path)
        print(f"✅ Nuevo mejor modelo guardado: {best_model_path} (Val Acc: {val_acc:.4f})")

    if early_stopper.early_stop:
        print("⛔ Entrenamiento detenido por EarlyStopping.")
        model.load_state_dict(early_stopper.best_model_wts)
        break

# ───────────────────────── Guardar historial ──────────────────────────
with open(pkl_path, "wb") as f:
    pickle.dump({
        'train_loss':      train_losses,
        'val_loss':        val_losses,
        'train_accuracy':  train_accuracies,
        'val_accuracy':    val_accuracies
    }, f)

# ──────────────────────── Resumen final ───────────────────────────────
print("\n✅ Final Summary:")
print(f"   Train Accuracy: {train_accuracies[-1]:.4f} | "
      f"Train Loss: {train_losses[-1]:.4f} | "
      f"Val Accuracy: {val_accuracies[-1]:.4f} | "
      f"Val Loss: {val_losses[-1]:.4f}")



```

```python
import os
import torch
import timm
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# ───── Configuración ─────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model_path = "/content/drive/MyDrive/Capstone/modelos/mejor_modelo_efficientvit_b1.pt"
val_dir = "/content/drive/MyDrive/Capstone/dataset/val"
img_size = 224
batch_size = 32
num_classes = 4

# ───── Transformaciones ─────
val_tfms = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

val_ds = datasets.ImageFolder(val_dir, transform=val_tfms)
val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False)
class_names = val_ds.classes

# ───── Cargar modelo entrenado ─────
model = timm.create_model(
    "efficientvit_b1",
    pretrained=False,
    num_classes=num_classes
)
model.load_state_dict(torch.load(model_path, map_location=device))
model.to(device)
model.eval()

# ───── Predicciones ─────
all_preds, all_labels = [], []
with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device)
        outputs = model(images)
        preds = torch.argmax(outputs, dim=1).cpu().numpy()
        all_preds.extend(preds)
        all_labels.extend(labels.numpy())

# ───── Matriz de Confusión ─────
cm = confusion_matrix(all_labels, all_preds)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - EfficientViT-B1")
plt.tight_layout()
plt.savefig("/content/drive/MyDrive/Capstone/modelos/matriz_confusion_efficientvit_b1.png", dpi=300)
plt.show()

# ───── Reporte de Clasificación ─────
report = classification_report(all_labels, all_preds, target_names=class_names)
print(report)

with open("/content/drive/MyDrive/Capstone/modelos/reporte_clasificacion_efficientvit_b1.txt", "w") as f:
    f.write(report)

```
